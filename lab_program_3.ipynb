```python
import pandas as pd

df = pd.read_csv('students_dataset.csv')
print(df.head())
```

### Distribution of Age (Histogram and Box Plot)

These plots help visualize the distribution of ages in the dataset and identify any potential outliers.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select the column for plotting
column_to_plot = 'Ageyears'

# Create a figure with two subplots
plt.figure(figsize=(14, 6))

# Subplot 1: Histogram
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
sns.histplot(df[column_to_plot].dropna(), kde=True)
plt.title(f'Histogram of {column_to_plot}')
plt.xlabel(column_to_plot)
plt.ylabel('Frequency')

```

```python
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
sns.boxplot(y=df[column_to_plot].dropna())
plt.title(f'Box Plot of {column_to_plot}')
plt.ylabel(column_to_plot)

plt.tight_layout()
plt.show()
```

### Distribution of Gender (Countplot)

This countplot visualizes the frequency of each gender category in the dataset.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.countplot(x='Gender', data=df, palette='viridis')
plt.title('Countplot of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

### Scatterplot of Age vs. Height

This scatterplot visualizes the relationship between the 'Ageyears' and 'Height_cm' to identify any correlation or trends.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))
sns.scatterplot(x='Ageyears', y='Height_cm', data=df)
plt.title('Scatterplot of Age vs. Height (cm)')
plt.xlabel('Age (Years)')
plt.ylabel('Height (cm)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Correlation Heatmap

This heatmap visualizes the correlation matrix between the numerical features in the dataset, helping to identify strong positive or negative relationships.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select only numerical columns for correlation calculation
numerical_cols = df.select_dtypes(include=['number']).columns

# Calculate the correlation matrix
correlation_matrix = df[numerical_cols].corr()

# Plot the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Numerical Features')
plt.show()
```
